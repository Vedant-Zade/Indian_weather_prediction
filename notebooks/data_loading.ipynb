```python
from google.colab import drive
drive.mount('/content/drive')

import xarray as xr
import os

BASE_PATH = "/content/drive/MyDrive/Indian Weather Prediction/datasets"

VARIABLES = [
    "t2m",
    "d2m",
    "sp",
    "msl",
    "u10",
    "v10",
    "tp"
]

def load_year(year):
    year_path = os.path.join(BASE_PATH, str(year))

    ds_inst = xr.open_dataset(os.path.join(year_path, "data_stream-oper_stepType-instant.nc"))
    ds_acc  = xr.open_dataset(os.path.join(year_path, "data_stream-oper_stepType-accum.nc"))

    ds = xr.merge([ds_inst, ds_acc])
    ds = ds[VARIABLES]

    return ds

# sanity check
ds_2019 = load_year(2019)
print(ds_2019)
```
